### **Step 2: Add More Tickets**

```python
def Add_More_Tickets():
    while True:
        ticket = int(input(f'How many new tickets do you want to add? '))
        if ticket <= 0:
            print(f'Invalid input: Please Enter a Positive number.')
            continue
        for i in range(ticket):
            ticket_data['Ticket_No'].append(len(ticket_data['Ticket_No'])+1)
            ticket_data['Customer_Name'].append(input(f'Enter the Customer Name: '))
            ticket_data['Issue_Description'].append(input('Enter the issue Description: '))
            while True:
                priority = input('Enter the Priority level: ')
                if priority == 'High' or priority == 'Medium' or priority == 'Low':
                    ticket_data['Priority'].append(priority)
                    break
                else:
                    print('Invalid input: Please Enter Hight, Medium or Low.')
                    continue
        print('New tickets added successfully!')
        break
```

```python
Add_More_Tickets()
```

```python
print(ticket_data)
```

### **Step 3: Text Cleaning for Issue Descriptions**

```python
ticket_data['Issue_Description']
```

```python
issues = ticket_data['Issue_Description']
cleaned_issues = []

for issue in issues:
    issue = issue.lower()  # Convert to Lowercase
    for x in '!,.?-':
        issue = issue.replace(x,'')  # Remove Punctuation
    issue = ' '.join(issue.split())  # Remove Extra Spaces
    issue = issue.strip()  # Remove Leading and Trailing Spaces
    cleaned_issues.append(issue)

ticket_data['Issue_Description'] = cleaned_issues
print(ticket_data['Issue_Description'])
```

```python
ticket_data['Issue_Description']
```

### **Step 4: Keyword-Based Issue Insights**

```python
def count_tickets_with_word(word):
    count = 0
    for issue in ticket_data['Issue_Description']:
        if word.lower() in issue.lower():
            count += 1
    return count
```

```python
count_tickets_with_word('PoOr')
```

```python
print('Number of tickets containing the word "poor":', count_tickets_with_word('poor'))
print('Number of tickets containing the word "good":', count_tickets_with_word('good'))
print('Number of tickets containing the word "slow":', count_tickets_with_word('slow'))
print('Number of tickets containing the word "excellent":', count_tickets_with_word('excellent'))
```

## **Step 5: Final Summary & Insights**

#### **1. Display Final Cleaned Ticket Data**

#### **2. Priority Analysis**

```python
high = ticket_data['Priority'].count('High')
medium = ticket_data['Priority'].count('Medium')
low = ticket_data['Priority'].count('Low')

print(f'Number of High Priority Tickets: {high}')
print(f'Number of Medium Priority Tickets: {medium}')
print(f'Number of Low Priority Tickets: {low}')
```

#### **3. Find the Ticket With the Longest Issue Description**

```python
max_words = 0
index = 0

for i, issue in enumerate(ticket_data["Issue_Description"]):
    words = len(issue.split())

    if words > max_words:
        max_words = words
        index = i

print("\nTicket With Longest Issue Description: ")
print("Ticket Number :", ticket_data["Ticket_No"][index])
print("Customer Name :", ticket_data["Customer_Name"][index])
print("Issue         :", ticket_data["Issue_Description"][index])
print("Word Count    :", max_words)
```

### **4. Extract Unique Words Used**

```python
unique_words = set()

for issue in ticket_data["Issue_Description"]:
    for word in issue.split():
        unique_words.add(word)

sorted_words = sorted(unique_words)

print("\nUnique Words Analysis")
print("Total Unique Words:", len(sorted_words))
print("Word List:")
print(sorted_words)
```
